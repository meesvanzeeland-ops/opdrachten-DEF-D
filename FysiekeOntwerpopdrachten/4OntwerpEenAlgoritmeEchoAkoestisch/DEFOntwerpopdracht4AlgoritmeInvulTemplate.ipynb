## Opdracht 3: Kalibratie sensor en foto

kalibratie-opstelling groep [38]

```python
# verwerk hier jullie kalibratiemetingen
A_bits = np.array([1837, 1537, 1306, 997, 857, 717, 550, 400, 1700, 1428, 1140]) #Arduino bits
Afstand = 100*np.array([0.6, 0.5, 0.4, 0.3, 0.25, 0.2, 0.15, 0.1, 0.55, 0.45, 0.35]) #Centimeter

def functie_fit(x, a, b):
    return a * x + b

values, cov = curve_fit(functie_fit, A_bits, Afstand)

x_functie_fit = np.linspace(400, 1900, 1000)
plt.title("Kalibratiegrafiek")
plt.plot(A_bits, Afstand, "k.", label="Meetpunten")
plt.xlabel("Arduino bits")
plt.ylabel("Afstand in centimeter $s$ [cm]")
plt.grid()
plt.plot(x_functie_fit, functie_fit(x_functie_fit, values[0], values[1]), "r--", label =r"functie fit volgens $y = a \cdot x + b$")
plt.legend()
plt.show()

print(values[0])
print(values[1])
```
