## *Opdracht 6*: Meten en fitten, Leerdoel

```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.optimize import curve_fit

m = 0.121 + np.array([0, 0.0463, 0.085, 0.143, 0.216, 0.250, 0.290, 0.345, 0.400, 0.500, -0.069])
Out = np.array([40, 58, 65, 82, 96, 99, 104, 118, 131, 150, 24])

plt.errorbar(m, Out,5,0.0001, "k.", label = "Metingen")
plt.xlabel("massa (kg)")
plt.ylabel("Output (Arduino bits)")
plt.legend()
```

```python

def weegschaal(a, x, b):
    return a*x +b
params, cov = curve_fit(weegschaal, m,Out, )
m_fit = np.linspace(0, 0.7, 100)
Out_fit = m_fit * params[0] + params[1]

plt.errorbar(m, Out,10,0.0001, "k.", label = "Metingen")
plt.plot(m_fit, Out_fit, "r--", label = "Fit")
plt.ylim(0,)
plt.xlim(0,)
plt.xlabel("massa (kg)")
plt.ylabel("Output (Arduino bits)")
plt.legend()
plt.show()
```
